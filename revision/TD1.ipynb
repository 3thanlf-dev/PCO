```python
def f2(a,b,c):
    d = b*b-4*a*c
    if(d>0):
        x1 = (-b + d**0.5)/(2*a)
        x2 = (-b - d**0.5)/(2*a)
        return (x1,x2)
    elif(d==0):
        x = -b/(2*a)
        return (x,)
    else:
        return ()
```
