```python
class point(object):
    def __init__ (self,x,y):
        self.x = x
        self.y = y

    def get_x(self):
        return self.x
    def set_x(self,x):
        self.x = x
        
    def get_y(self):
        return self.y

    def set_y(self,y):
        self.y = y

    def __add__(self,truc):
        som1 = (self.x + truc.x)
        som = (self.y + truc.y)
        return point(som1,som)

    def __mul__(self,autre):
        mul1 = (self.x * autre.x)
        mul = (self.y * autre.y)
        return point(mul1,mul)

    
    def __str__(self):
        return str(self.x) + ";;;;" +  str(self.y)
    
```

```python
class segment(object):
    def __init__ (self,x,y):
        self.origin = x
        self.target = y

    def affiche(self):
        import pylab
        pylab.plot([self.origin.x,self.target.x],[self.origin.y,self.target.y])
        pylab.show()

    def longueur(self):
        l=((self.origin.x-self.target.x)*(self.origin.x-self.target.x)+(self.origin.y-self.target.y)*(self.origin.y-self.target.y))**1/2
        return l

    def translation(self,vecteur):
        originnew = (self.origin + vecteur)
        targetnew = (self.target + vecteur) 
        return segment(originnew, targetnew)
        
```
