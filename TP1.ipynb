```python
#Exemple du cours (CM2)
class Coordinate(object):
    """ A coordinate made up of an x and y value """
    def __init__(self, x, y):
        """ Sets the x and y values """
        self.x = x
        self.y = y
    def __str__(self):
        """ Returns a string representation of self """
        return "<" + str(self.x) + "," + str(self.y) + ">"
    def distance(self, other):
        """ Returns the euclidean distance between two points """
        x_diff_sq = (self.x-other.x)**2
        y_diff_sq = (self.y-other.y)**2
        return (x_diff_sq + y_diff_sq)**0.5
```

```python
c = Coordinate(3,4)
origin = Coordinate(0,0)
print(c.x, origin.x)
print(c.distance(origin))
print(Coordinate.distance(c, origin))
print(origin.distance(c))
print(c)
```

```python
class Fraction(object):
    """
    une fraction est représentée par un numerateur(num) et dénominateur(demon)
    """
    def __init__(self, num, denom):
        """ num et denom sont des entiers """
        assert type(num) == int and type(denom) == int 
        "possibilité avec des assertions de vérifier le type de num et denom"
        self.num = num
        self.denom = denom
    def __str__(self):
        """ Retourne une chaine pour représenter self, utilisé lors de l'appel de print """
        return str(self.num) + "/" + str(self.denom)
    def __add__(self, other):
        """ Retourne une nouvelle fraction résultat de l'addition """
        top = self.num*other.denom + self.denom*other.num
        bott = self.denom*other.denom
        return Fraction(top, bott)
    def __sub__(self, other):
        """  Retourne une nouvelle fraction résultat de la soustraction """
        top = self.num*other.denom - self.denom*other.num
        bott = self.denom*other.denom
        return Fraction(top, bott)
    def __float__(self):
        """Retourne une valeur flottante pour la fraction """
        return self.num/self.denom
    def inverse(self):
        """  Retourne une nouvelle fraction representaant l'inverse 1/self """
        return Fraction(self.denom, self.num)
```

```python
a = Fraction(1,4)
b = Fraction(3,4)
c = a + b # c is a Fraction object
print(c)
print(float(c))
print(Fraction.__float__(c))
print(float(b.inverse()))
#c = Fraction(3.14, 2.7) # assertion error
#print a*b # error, did not define how to multiply two Fraction objects
```

```python
class CompteSimple(object):
    def __init__(self,solde,titulaire):
        self.solde=solde
        self.titulaire=titulaire
    def get_solde(self):
        return self.solde
    def get_titulaire(self):
        return self.titulaire
    def set_solde(self, solde):
        self.solde=solde
    def set_titulaire(self,titulaire):
        self.titulaire=titulaire
    def crediter (self, montant):
        self.set_solde(montant + self.get_solde())
    def débiter (self, montant):
        self.set_solde(montant - self.get_solde())

c1=CompteSimple(100,None)
c1.crediter(50)
print(c1.get_solde())
print(c1.get_titulaire())


class Personne(object):
    """Une personne n'est pas une sous-classe de CompteSimple
    """
    def __init__(self, identité, adresse):
        self.identité= identité
        self.adresse=adresse
    def get_identité(self):
        return self.identité
    def get_adresse(self):
        return self.adresse
    def set_identité(self, identité):
        self.identité=identité
    def set_adresse(self, adresse):
        self.adresse=adresse
    def __str__(self):
        return str(self.get_identité()) + "@" + str(self.get_adresse())
    
toto=Personne("Paul_Dupont", "Brest")
print( toto.get_identité())
print( toto.get_adresse())
print(toto)

c1.set_titulaire(toto)
print(c1.get_titulaire())
```

```python
class Point (object):
    def __init__(self, x=0, y=0):
        self.x = x
        self.y = y
    def get_x(self):
        return self.x
    def get_y(self):
        return self.y
    def set_x(self, x):
        self.x=x
    def set_y(self,y):
        self.y=y
    
    def __add__(self, other):
        return Point (self.x + other.x, self.y +other.y)

    def __str__(self):
         return "{" + str(self.x) + "," + str(self.y) + "}"
        
    def translater(self, dx, dy):
        "version 1"
        "Cette méthode retourne self par defaut qui est modifié avec les nouvelles valeurs "
        self.x += dx
        self.y += dy
        
    def translaterP(self, point):
        "version 2"
        "Cette méthode retourne un nouveau point et utiliser l'addition sur les point "
        newpoint= self + point
        return newpoint
        
        
    def distance(self, autre):
        dx2 = (self.x - autre.x) ** 2
        dy2 = (self.y - autre.y) ** 2
        return (dx2 + dy2)**0.5
    
p1= Point()
print("point p1", p1)
p2= Point(3,4)
print("point p2",p2)
p3=p2+p2
print("addition de points p2+p2=p3", p3)
v1=Point(1,2)
p4=p2.translaterP(v1)
print("nouveau point  apres translation (version2) de p2: p4",p4)
print("p2 n'est pas modifié",p2)
p2.translater(1,2) #p2 est modifié par la translation
print("Nouvelle valeur de  p2 modifié après translation (version1)",p2)
print(p2)
print("La distance entre p3 et p4 est de ", p3.distance(p4))
```
