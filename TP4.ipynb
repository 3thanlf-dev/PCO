```python
class NodePondere(object):
    def __init__(self,numero, poids):
        self.numero = numero
        self.poids = poids
    def getName(self):
        return str(self.numero)
    def get_numero(self):
        return self.numero
    def get_poids(self):
        return self.poids
    def set_numero(self, new_numero):
        self.numero = new_numero
    def set_poids(self, new_poids):
        self.poids = new_poids
    def __str__(self):
        return "("+str(self.numero)+","+str(self.poids)+")"

```

```python
listPoids = [50,13,20,35,5,70,53,80,13,1]
listNP = []
for i in range(0,10):
    listNP.append(NodePondere(i,listPoids[i]))
    print(listNP[i])
    
```

```python
class Node(object):
    def __init__(self,name):
        self.name= name
    def getName(self):
        return self.name
    def __str__(self):
        return self.name
    
class Edge(object):
    def __init__(self, src, dest):
        self.src=src
        self.dest=dest
    def getSource(self):
        return self.src;
    def getDestination(self):
        return self.dest
    def __str__(self):
        return self.src.getName + '->' + self.dest.getname()

class Digraph(object):
    #nodes est une liste de noeuds
    #edges est un dictionnaire mappant chaque noeud à ses enfants
    def __init__(self):
        self.nodes=[]
        self.edges={}
    def addNode(self, node):
        if node in self.nodes:
            raise ValueError('Duplicate node')
        else:
            self.nodes.append(node)
            self.edges[node]=[]
    def addEdge (self, edge):
        src = edge.getSource()
        dest= edge.getDestination()
        if not(src in self.nodes and dest in self.nodes):
            raise ValueError('Node not in graph')
        self.edges[src].append(dest)
    def childrenOf(self, node):
        return self.edges[node]
    
    def hasNode(self,node):
        return node in self.nodes
    def __str__(self):
        result=''
        for src in self.nodes:
            for dest in self.edges[src]:
                result=result + src.getName() + '->'\
                        + dest.getName() + '\n'
        return result[:-1]
    
nodes=[]
for name in range(6): #6 noeuds
        nodes.append(Node(str(name)))
        
g=Digraph()
for n in nodes:
      g.addNode(n)
g.addEdge(Edge(nodes[0], nodes[1]))
g.addEdge(Edge(nodes[1], nodes[2]))
g.addEdge(Edge(nodes[2], nodes[3]))
g.addEdge(Edge(nodes[2], nodes[4]))
g.addEdge(Edge(nodes[3], nodes[4]))
g.addEdge(Edge(nodes[3], nodes[5]))
g.addEdge(Edge(nodes[0], nodes[2]))
g.addEdge(Edge(nodes[1], nodes[0]))
g.addEdge(Edge(nodes[3], nodes[1]))
g.addEdge(Edge(nodes[4], nodes[0]))
print(g) 
```

```python
def printPath(path):
    result=''
    for i in range(len(path)):
        result= result + str(path[i])
        if i != len(path) -1:
            result =result + '->'
    return result

def DFS(graph, start, end, path, all_paths):
    path = path + [start]
    print('Current DFS path', printPath(path))
    if start == end:
        all_paths.append(path)
        return all_paths
    for node in graph.childrenOf(start):
        if node not in path: 
            DFS(graph, node, end, path, all_paths)
    return all_paths

def search(graph, start, end):
    return DFS(graph, start, end, [], [])

chemins = search(g_np, listNP[0], listNP[5])
print('\nTous les chemins trouvés :')
for i, chemin in enumerate(chemins, 1):
    print(f'{i}. {printPath(chemin)}')
```

```python
def BFS(graph, start, end):
    initPath=[start]
    pathQueue=[initPath]
    while len(pathQueue) !=0:
        tmpPath = pathQueue.pop(0)
        print('Current BFS path:', printPath(tmpPath))
        lastNode= tmpPath[-1]
        if lastNode==end:
            return tmpPath
        print('lastNode', lastNode)
        for nextNode in graph.childrenOf(lastNode):
            if nextNode not in tmpPath:
                print('Node', nextNode)
                newPath = tmpPath + [nextNode]
                pathQueue.append(newPath)
                
    return pathQueue



sp=BFS(g_np, listNP[0], listNP[5])
print('Chemin le plus court avec  BFS', printPath(sp))
```
